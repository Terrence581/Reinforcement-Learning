```python
def k_step_belief(k, belief, tran_0, tran_1):
    """
    Function:
    - Compute the k-step belief state.

    Input:
    - k: a scalar indicating whose belief state the function computes
    - belief: a scalar or an array indicating current belief state
    - tran_0: a scalar or an array showing the transition probablity p_{01}
    - tran_1: a scalar or an array showing the transition probablity p_{11}
    """
    temp = tran_1 - tran_0
    return (tran_0 - (temp**k)*(tran_0 - (1 -temp)*belief))/(1 - temp)


def fir_cross_t(cur, tar, tran_0, tran_1, ini_belief):
    """
    Function:
    - Compute the first crossing time, from current belief state to target belief state, for positive correlated arm.

    Input:
    - cur: a scalar or an array showing the current belief state
    - tar: a scalar or an array showing the target belief state
    - ini_belief: a scalar or an array showing the initial belief state w_{0}
    """
    L = np.zeros(np.size(cur))
    L[(cur <= tar)*(tar >= ini_belief)] = float('inf')
    temp_idx = (cur <= tar)*(tar < ini_belief)
    temp_val = 1 - tran_1[temp_idx] + tran_0[temp_idx]
    num = tran_0[temp_idx] - tar[temp_idx]*temp_val
    deno = tran_0[temp_idx] - cur[temp_idx]*temp_val
    L[temp_idx] = 1 + (np.log(num/deno)/np.log(1 - temp_val)).astype(int)
    # print(L)
    return L


```

```python
def whittle_idx_pos(w, beta, reward, tran_0, tran_1, ini_belief):
    """
    Function:
    - Compute the whittle index for positive correlated arm(channel)

    Input:
    - w: a scalar or an array showing the belief state for the whittle index
    - beta: a scalar indicating the discounted factor
    - reward: a scalar or an array showing the reward of good state
    """

    whittle_idx = np.zeros(np.size(w))

    idx1 = (w <= tran_0) + (w >= tran_1)
    idx2 = (ini_belief <= w)*(w < tran_1)
    idx3 = (tran_0 < w)*(w < ini_belief)

    L = fir_cross_t(tran_0[idx3], w[idx3], tran_0[idx3], tran_1[idx3], ini_belief[idx3])
    T_L = k_step_belief(L, tran_0[idx3], tran_0[idx3], tran_1[idx3])
    beta_L = beta**(L + 1)
    temp1 = 1 - beta*tran_1[idx3]
    temp2 = temp1*(1 - beta_L) + (1 - beta)*beta_L*T_L
    C1 = temp1*(1 - beta_L/beta)/temp2
    C2 = T_L*(beta_L/beta)/temp2
    T_1 = k_step_belief(1, w[idx3], tran_0[idx3], tran_1[idx3])
    temp3 = w[idx3] - beta*T_1
    temp4 = beta*(temp1 - w[idx3] - temp3)

    whittle_idx[idx1] = w[idx1]*reward[idx1]
    whittle_idx[idx2] = (w[idx2]*reward[idx2])/(1 - beta*tran_1[idx2] + beta*w[idx2])
    whittle_idx[idx3] = reward[idx3]*(temp3 + C2*(1 - beta)*temp4)/(temp1 - C1*temp4)

    return whittle_idx


def whittle_idx_neg(w, beta, reward, tran_0, tran_1, ini_belief):
    """
    Function:
    - Compute the whittle index for negative correlated arm(channel)
    """

    whittle_idx = np.zeros(np.size(w))

    T_p = k_step_belief(1, tran_1, tran_0, tran_1)
    idx1 = (w <= tran_1) + (w >= tran_0)
    idx2 = (T_p <= w)*(w < tran_0)
    idx3 = (ini_belief <= w)*(w < T_p)
    idx4 = (tran_1 < w)*(w < ini_belief)

    temp1_3 = 1 + (1 + beta)*beta*tran_0[idx3] - T_p[idx3]*(beta**2)
    temp1_4 = 1 + (1 + beta)*beta*tran_0[idx4] - T_p[idx4]*(beta**2)
    C3_3 = (1 - beta*(1 - tran_0[idx3]))/temp1_3
    C3_4 = (1 - beta*(1 - tran_0[idx4]))/temp1_4
    C4_3 = (beta*T_p[idx3]*(1 - beta) + tran_0[idx3]*(beta**2))/temp1_3
    C4_4 = (beta*T_p[idx4]*(1 - beta) + tran_0[idx4]*(beta**2))/temp1_4
    temp2 = beta*tran_0[idx3] + w[idx3]*(1 - beta)
    temp3_3 = 1 - beta*(1 - tran_0[idx3])
    temp3_4 = 1 - beta*(1 - tran_0[idx4])
    T_1 = k_step_belief(1, w[idx4], tran_0[idx4], tran_1[idx4])
    temp4 = beta*T_1 - beta*tran_0[idx4] - w[idx4]

    whittle_idx[idx1] = w[idx1]*reward[idx1]
    whittle_idx[idx2] = reward[idx2]*(beta*tran_0[idx2] + w[idx2]*(1 - beta))/(1 + beta*(tran_0[idx2] - w[idx2]))
    whittle_idx[idx3] = reward[idx3]*(1 - beta + beta*C4_3)*temp2/(temp3_3 - C3_3*beta*temp2)
    whittle_idx[idx4] = reward[idx4]*((beta - 1)*temp4 - C4_4*beta*temp4)/(temp3_4 + C3_4*beta*temp4)

    return whittle_idx


```

```python
def dis_val_pos(w, beta, w_opt, m, tran_0, tran_1, ini_belief):
    """
    Function:
    - Compute the exptected total discounted reward for positive correlated arm(channel).

    Input:
    - m: a scalar indicating the subsidy for passive arm(channel)
    """

    V_w = np.zeros(np.size(w))
    V_p0 = np.zeros(np.size(w))
    V_p1 = np.zeros(np.size(w))

    idx1 = (w_opt < tran_0)
    idx2 = (tran_0 <= w_opt)*(w_opt < ini_belief)
    L1 = fir_cross_t(tran_0[idx2], w_opt[idx2], tran_0[idx2], tran_1[idx2], ini_belief[idx2])
    T1 = k_step_belief(L1, tran_0[idx2], tran_0[idx2], tran_1[idx2])
    temp1 = 1 - beta*tran_1[idx2]
    temp2 = beta**L1
    temp3 = (1 - beta)*temp3*T1

    V_p0[idx1] = tran_0[idx1]/((1- beta)*(1 - beta*tran_1 + beta*tran_0))
    V_p0[w_opt >= ini_belief] = m/(1 - beta)
    V_p0[idx2] = (temp1*(1 - temp2)*m + temp3)/(temp1*(1 - beta)*(1 - temp2*beta) + (1 - beta)*beta*temp3)

    idx3 = (w_opt < tran_1)

    V_p1[idx3] = (tran_1[idx3] + beta*(1 - tran_1[idx3])*V_p0[idx3])/(1 - beta*tran_1[idx3])
    V_p1[w_opt >= tran_1] = m/(1 - beta)

    idx4 = (w == tran_0)
    idx5 = (w == tran_1)
    idx6 = (w != tran_0)*(w != tran_1)
    L2 = fir_cross_t(w[idx6], w_opt[idx6], tran_0[idx6], tran_1[idx6], ini_belief[idx6])
    T2 = k_step_belief(L2, w[idx6], tran_0[idx6], tran_1[idx6])
    temp4 = beta**L2

    V_w[idx4] = V_p0[idx4]
    V_w[idx5] = V_p1[idx5]
    V_w[idx6] = m*(1 - temp4)/(1 - beta) + temp4*(T2 + T2*beta*V_p1[idx6] + beta*(1 - T2)*V_p0[idx6])

    return V_w

def dis_val_neg(w, beta, w_opt, m, tran_0, tran_1, ini_belief):
    """
    Function:
    - Compute the exptected total discounted reward for negative correlated arm(channel).
    """

    V_w = np.zeros(np.size(w))
    V_p0 = np.zeros(np.size(w))
    V_p1 = np.zeros(np.size(w))

    idx1 = (w_opt < tran_1)
    T1 = tran_1*tran_1 + (1 - tran_1)*tran_0
    idx2 = (tran_1 <= w_opt)*(w_opt < T1)
    temp1 = 1 - beta*(1 - tran_0[idx2])
    temp2 = beta*(T1[idx2]*(1 - beta) + beta*tran_0[idx2])

    V_p1[idx1] = (tran_1[idx1]*(1 - beta) + beta*tran_0[idx1])/((1 - beta)*(1 - beta*tran_1[idx1] + beta*tran_0[idx1]))
    V_p1[w_opt >= T1] = m/(1 - beta)
    V_p1[idx2] = (m*temp1 + temp2)/(temp1 - beta*temp2)

    idx3 = (w_opt < tran_0)
    V_p0[idx3] = (tran_0[idx3] + beta*tran_0[idx3]*V_p1[idx3])/(1 - beta*(1 - tran_0[idx3]))
    V_p0[w_opt >= tran_0] = m/(1 - beta)

    T2 = w*tran_1 + (1 - w)*tran_0
    idx4 = (w == tran_0)
    idx5 = (w == tran_1)
    idx6 = (w != tran_0)*(w != tran_1)
    idx7 = idx6*(w > w_opt)
    idx8 = idx6*(w <= w_opt)*(T2 > w_opt)
    
    V_w[idx4] = V_p0[idx4]
    V_w[idx5] = V_p1[idx5]
    V_w[idx7] = w[idx7] + beta*(w[idx7]*V_p1[idx7] + (1 - w[idx7])*V_p0[idx7])
    V_w[idx6*(w <= w_opt)*(T2 <= w_opt)] = m/(1 -beta)
    V_w[idx8] = m + beta*(T2[idx8] + T2[idx8]*beta*V_p1[idx8] + beta*(1 - T2[idx8])*V_p0[idx8])

    return V_w
```

```python
def dis_pass_time_pos(w, beta, w_opt, tran_0, tran_1, ini_belief):
    """
    Function:
    - Compute the exptected total discounted passive time for positive correlated arm(channel).
    """
    D_w = np.zeros(np.size(w))
    D_p0 = np.zeros(np.size(w))
    D_p1 = np.zeros(np.size(w))

    idx1 = (tran_0 <= w_opt)*(w_opt < ini_belief)
    temp1 = 1 - beta*tran_1[idx1]
    L1 = fir_cross_t(tran_0[idx1], w_opt[idx1], tran_0[idx1], tran_1[idx1], ini_belief[idx1])
    T1 = k_step_belief(L1, tran_0[idx1], tran_0[idx1], tran_1[idx1])
    temp1 = 1 - beta*tran_1[idx1]
    temp2 = beta**(L1 + 1)

    D_p0[w_opt < tran_0] = 0
    D_p0[w_opt >= ini_belief] = 1/(1 - beta)
    D_p0[idx1] = temp1*(1 - temp2/beta)/(temp1*(1 - beta)*(1 - temp2) + temp2*T1*((1 - beta)**2))

    idx2 = (w_opt < tran_1)

    D_p1[idx2] = beta*(1 - tran_1[idx2])*D_p0[idx2]/(1 - beta*tran_1[idx2])
    D_p1[w_opt >= tran_1] = 1/(1 - beta)

    idx3 = (w == tran_0)
    idx4 = (w == tran_1)
    idx5 = (w != tran_0)*(w != tran_1)
    L2 = fir_cross_t(w[idx5], w_opt[idx5], tran_0[idx5], tran_1[idx5], ini_belief[idx5])
    T2 = k_step_belief(L2, w[idx5], tran_0[idx5], tran_1[idx5])
    temp3 = beta**L2

    D_w[idx3] = D_p0[idx3]
    D_w[idx4] = D_p1[idx4]
    D_w[idx5] = (1 - temp3)/(1 - beta) + beta*temp3*(T2*D_p1[idx5] + (1 - T2)*D_p0[idx5])

    return D_w


def dis_pass_time_neg(w, beta, w_opt, tran_0, tran_1, ini_belief):
    """
    Function:
    - Compute the exptected total discounted passive time for negative correlated arm(channel).
    """
    D_w = np.zeros(np.size(w))
    D_p0 = np.zeros(np.size(w))
    D_p1 = np.zeros(np.size(w))

    T1 = tran_1*tran_1 + (1 - tran_1)*tran_0
    idx1 = (tran_1 <= w_opt)*(w_opt < T1)
    temp1 = 1 - beta*(1 - tran_0[idx1])

    D_p1[(w_opt < tran_1)] = 0
    D_p1[w_opt => T1] = 1/(1 - beta)
    D_p1[idx1] = temp1/(temp1 - T1[idx1]*(1 - beta)*(beta**2) - tran_0[idx1]*(beta**3))

    idx2 = (w_opt < tran_0)

    D_p0[idx2] = beta*tran_0[idx2]*D_p1[idx2]/(1 - beta*(1 - tran_0[idx2]))
    D_p0[w_opt >= tran_0] = 1/(1 - beta)

    idx3 = (w == tran_0)
    idx4 = (w == tran_1)
    idx5 = (w != tran_0)*(w != tran_1)
    T2 = w*tran_1 + (1 - w)*tran_0
    idx6 = idx5*(w <= w_opt)*(T2 > w_opt)
    idx7 = idx5*(w > w_opt)
    T3 = k_step_belief(1, w[idx6], tran_0[idx6], tran_1[idx6])

    D_w[idx3] = D_p0[idx3]
    D_w[idx4] = D_p1[idx4]
    D_w[idx5*(w <= w_opt)*(T2 <= w_opt)] = 1/(1 - beta)
    D_w[idx6] = 1 + (beta**2)*(T3*D_p1[idx6] + (1 - T3)*D_p0[idx6])
    D_w[idx7] = beta*(w[idx7]*D_p1[idx7] + ( 1 - w[idx7])*D_p0[idx7])

    return D_w



```

```python
def find_w_opt(ini_belief, delta, arm_num, beta, tran_0, tran_1, reward, lower_bound, target, level=1):
    """
    Function:
    - Find the optimal belief state for positive corrlated arm(channel)

    Input:
    - arm_num: a scalar indicating the number of arms (channels)
    - delta: a scalar indicating the total length of gray area over all channels
    - lower_bound: the lower bound for the dichotomy
    - target: the target at which the whittle index for optimal belief state should arrive
    - level: the level of interval length
    """
    w_num = np.size(ini_)
    w_opt = np.zeros(w_num) - 1
    length = delta/arm_num
    if (2**(-level)) < lower_bound:
        return w_opt

    for j in range(1, 2**(level), 2):
        whittle_com = np.zeros(np.size(target)) - 100
        w = np.zeros(np.size(w_opt))
        need_idx = (w_opt == -1)
        w[need_idx] = ini_belief[need_idx] - length*(2**(-level))*j
        whittle_com[need_idx] = whittle_idx_pos(
            w, beta, reward[need_idx], tran_0[need_idx], tran_1[need_idx], ini_belief[need_idx])
        opt_idx = (whittle_com <= target)
        w_opt[opt_idx] = w[opt_idx]

    if (w_opt != -1).all():
        return w_opt

    else:
        find_idx = (w_opt == -1)
        w_opt[find_idx] = find_w_opt(ini_belief[find_idx], delta, beta, tran_0[find_idx],tran_1[find_idx], reward[find_idx], error, target[find_idx], level + 1)
        return w_opt
```

```python
def find_li(target, w, tran_0, tran_1, ini_belief, beta, reward):
    h = 1
    #li = np.ones(np.size(tran_1))*(-1)
    whittle_Tliw_pos = np.array([])
    whittle_Tlw_pos = np.zeros(np.size(tran_1))

    T_h_w_pos = k_step_belief(h, w, tran_0, tran_1)
    need_idx = (T_h_w_pos <= target)
    while need_idx.any():
        #li[(li == -1)*(T_h_p0_pos > target)] = h
        whittle_Thw_pos = whittle_idx_pos(T_h_w_pos[need_idx], beta, reward[need_idx], tran_0[need_idx], tran_1[need_idx], ini_belief[need_idx])
        whittle_Tliw_pos = np.append(whittle_Tliw_pos, whittle_Thw_pos)
        h = h + 1
        T_h_w_pos[need_idx] = k_step_belief(h, w[need_idx], tran_0[need_idx], tran_1[need_idx])
        need_idx = (T_h_w_pos <= target)

    #li[li == -1] = h
    whittle_Tlw_pos = whittle_idx_pos(T_h_w_pos, beta, reward, tran_0, tran_1, ini_belief)
    whittle_Tliw_pos = np.append(whittle_Tliw_pos, whittle_Tlw_pos)

    return whittle_Tliw_pos, whittle_Tlw_pos
```
